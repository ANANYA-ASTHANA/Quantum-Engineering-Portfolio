## Key Ideas

### 1) Qubit Statevector
A single-qubit pure state is:
$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle,\quad |\alpha|^2 + |\beta|^2 = 1
$

- **Global phase** doesn’t matter physically  
- **Relative phase** \($\arg(\alpha)-\arg(\beta)$\) affects interference

### 2) Measurement
Measuring in the **Z basis** yields:
- Outcome 0 with probability $|\alpha|^2$ 
- Outcome 1 with probability $|\beta|^2$

Different basis measurements require **unitary rotations** before Z-measurement.

### 3) Bloch Sphere Representation
Any pure qubit state can be written as:
$
|\psi\rangle = \cos(\tfrac{\theta}{2})|0\rangle + e^{i\phi}\sin(\tfrac{\theta}{2})|1\rangle.
$

- $\theta$: population balance  
- $\phi$: relative phase  
- Useful for visual intuition and debugging

## Example Code — Prepare $|+\rangle$ and visualize

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)
qc.h(0)          # Prepare |+>
qc.measure_all()
qc.draw("mpl")
```

### Bloch vector of $|0\rangle$

```python
from qiskit.quantum_info import Statevector
Statevector.from_label("0").draw("bloch")
```
