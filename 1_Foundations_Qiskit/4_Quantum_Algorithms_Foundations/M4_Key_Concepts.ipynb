## Key Ideas

### 1) Interference
Superposed amplitudes combine:  
- Constructively $ \rightarrow $ high measurement probability  
- Destructively $ \rightarrow $ suppressed outcomes

### 2) Phase Kickback
Using CZ/CU operations, a phase applied to a **target** can “kick back” to the **control** qubit’s amplitude — useful for extracting properties of oracles.

### 3) Oracles
Reversible circuits implementing classical functions: $|x\rangle|y\rangle \mapsto |x\rangle |y \oplus f(x)\rangle$.  
Phases encode information compactly.

## Example Code — Minimal Kickback Sketch

```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.x(1)
qc.cz(0,1)
qc.h(0)
qc.measure_all()
qc.draw("mpl")
```
