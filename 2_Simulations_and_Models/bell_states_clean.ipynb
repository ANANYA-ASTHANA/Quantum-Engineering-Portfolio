# Bell States: Generation and Verification

## Objectives
- Prepare maximally entangled Bell states using minimal circuits.
- Verify entanglement via statevector fidelity and correlation checks.

## Setup
```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector, state_fidelity
import matplotlib.pyplot as plt
```


## Theory Snapshot
Bell states are maximally entangled two‑qubit states. We focus on $\displaystyle |\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$. Ideal simulators yield unit fidelity; hardware or noise models reduce visibility.

## Circuit / Model
We prepare $|\Phi^+\rangle$ with a single Hadamard on qubit 0 followed by a CNOT $(0\rightarrow 1)$.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, state_fidelity

def bell_phi_plus_circuit():
    qc = QuantumCircuit(2)
    qc.h(0); qc.cx(0,1)
    return qc

bell_qc = bell_phi_plus_circuit()
bell_qc
````

## Experiments
We evaluate fidelity of the prepared state against the target $|\Phi^+\rangle$ and sample measurement statistics in the computational basis.

````python
# Target statevector for |phi+>
import numpy as np
target = Statevector([1/np.sqrt(2),0,0,1/np.sqrt(2)])

# Prepare state from circuit
sv = Statevector.from_label('00').evolve(bell_qc)

fid = state_fidelity(sv, target)
print("Fidelity(|Φ+⟩):", float(fid))
````

````python
# Sample measurement statistics
from qiskit_aer import Aer
backend = Aer.get_backend('aer_simulator')
meas = bell_qc.copy(); meas.measure_all()
tq = backend.run(meas, shots=2000).result().get_counts()
tq
````

## Metrics & Plots
- **Fidelity:** similarity to $|\Phi^+\rangle$. Expect $\approx 1.0$ on ideal simulator.
- **Counts:** dominant outcomes 00 and 11 with $ \approx 50/50$ split.

(Optional) Plot correlations.

````python
try:
    import matplotlib.pyplot as plt
    zeros = tq.get('00',0); elev = tq.get('11',0); others = sum(v for k,v in tq.items() if k not in ('00','11'))
    plt.figure(); plt.bar(['00','11','others'], [zeros, elev, others])
    plt.title('Bell |Φ+⟩ measurement distribution'); plt.grid(True, axis='y')
except Exception as e:
    print("Plot skipped:", e)
````

## Results & Discussion
- The prepared state matches $|\Phi^+\rangle$ with near‑unit fidelity on ideal simulation; counts cluster on 00 and 11.
- Under noise, parity correlations degrade; mitigation includes calibration and error‑mitigation techniques.
