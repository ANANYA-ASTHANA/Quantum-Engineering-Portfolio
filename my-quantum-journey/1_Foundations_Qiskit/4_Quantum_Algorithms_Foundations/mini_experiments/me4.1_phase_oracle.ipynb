# ME4.1: Phase Oracle

## Objectives
- Implement a phase oracle; show phase kickback to a control register.
- Contrast global vs relative phase observability.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
```


## Theory Snapshot
- Oracles often add $(−1)^{f(x)}$ phase; Hadamards map phase patterns to bitstrings.
- Global phase cancels in measurement; relative phase changes interference.

## Experiment


````python
from qiskit import QuantumCircuit
from qiskit_aer import Aer

sim = Aer.get_backend('aer_simulator')

def p_plus(with_oracle, shots=1024):
    qc = QuantumCircuit(2, 1)
    qc.h(0)            # control: |+>
    qc.x(1)            # target: |1> (eigenstate of Z with eigenvalue -1)
    if with_oracle:
        qc.cz(0, 1)    # phase oracle: (-1) on |11>
    qc.h(0)            # measure control in X-basis (H -> Z, then measure)
    qc.measure(0, 0)
    counts = sim.run(qc, shots=shots).result().get_counts()
    return counts.get('0', 0) / shots  # '0' corresponds to |+> in X-basis

print("P(+), no oracle :", p_plus(False))
print("P(+), with oracle:", p_plus(True))
````

## Results & Discussion

### Results
- With Hadamard “sandwich”, oracle phase converts to deterministic bit outcomes.
- Removing the uncompute (in this case, the Hadamard) masks the phase (no signal).

### Discussion
- Clear evidence that only relative phase impacts measurement statistics.
- Uncomputation is crucial to read out oracle-encoded phase.
