## Key Ideas

### 1) Measurement Operators
Measurement is described by operators $\{M_k\}$ satisfying: $\sum_k M_k^\dagger M_k = I.$

- Projective measurement $ \rightarrow $ Hermitian projectors  
- POVMs $ \rightarrow $ generalized measurements

### 2) Basis Measurements
- **X basis:** $H  \rightarrow$  measure $Z$  
- **Y basis:** $S^\dagger \rightarrow  H  \rightarrow$  measure $Z$

### 3) Noise Channels
Examples:  
- **Bit flip:** $ \rho \mapsto (1-p)\rho + pX\rho X $  
- **Phase flip**  
- **Amplitude damping** (models relaxation)

Noise transforms pure states into mixed states and alters correlations.

## Example Code — Y-basis + Amplitude Damping

```python
from qiskit import QuantumCircuit
# Y-basis measurement of |1>
qc = QuantumCircuit(1)
qc.x(0)
qc.sdg(0); qc.h(0)
qc.measure_all()
qc.draw("mpl")
```

```python
from qiskit_aer.noise import NoiseModel, amplitude_damping_error
noise = NoiseModel()
noise.add_all_qubit_quantum_error(amplitude_damping_error(0.15), ["id", "rz"])
noise
```

## Summary
Measurement extracts classical information; noise corrupts it. Understanding basis rotations and channel effects is essential when interpreting real devices or simulators.
