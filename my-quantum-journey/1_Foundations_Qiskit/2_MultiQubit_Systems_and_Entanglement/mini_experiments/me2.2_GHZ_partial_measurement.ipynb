# ME2.2: GHZ Partial Measurement

## Objectives
- Prepare 3-qubit GHZ; measure one qubit; inspect conditional state of the rest.
- Demonstrate projection collapsing the remainder into a Bell-like state.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```


## Theory Snapshot
- $∣GHZ\rangle=(∣000\rangle+∣111\rangle)/\sqrt(2)$; Z-measuring one qubit projects the other two into $∣00\rangle$ or $∣11\rangle$ (up to phase).
- Conditioning on the outcome reveals strong two-qubit correlations.

## Experiment



````python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram

pm=QuantumCircuit(3,3) 
pm.h(0); pm.cx(0,1); pm.cx(0,2)

pm.measure(0,0) 
pm.barrier()
pm.measure([1,2],[1,2])
sim=Aer.get_backend('aer_simulator')
plot_histogram(sim.run(pm, shots=1024).result().get_counts())
````

## Results & Discussion

### Results
- Measuring qubit-0: outcomes {0,1} $\approx$ balanced.
- Conditional on $0\rightarrow$ others near $∣00\rangle$; on $1\rightarrow$ near $∣11\rangle$.

### Discussion
- Post-measurement entanglement structure matches GHZ projection theory.
- Correlation visibility limited only by finite shots and any inserted noise.
