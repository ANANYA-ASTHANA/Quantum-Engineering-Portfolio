# ME2.1: Bell in X/Y/Z Bases

## Objectives
- Prepare $|\Phi+\rangle$ and measure correlations in Z, X, and Y bases.
- Estimate $\langle ZZ\rangle,\langle XX\rangle,\langle YY\rangle$ qualitatively from histograms.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```


## Theory Snapshot
- $|\Phi+\rangle$ stabilized by $ZZ$ and $XX \rightarrow$ correlated in Z and X.
- $YY$=−1 for $|\Phi+\rangle \rightarrow$ anti-correlated in Y.

## Experiment



````python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram

qc=QuantumCircuit(2); qc.h(0); qc.cx(0,1)
sim=Aer.get_backend('aer_simulator')

def m(c,b='Z'):
    c=c.copy()
    if b=='X': c.h([0,1])
    elif b=='Y': c.sdg([0,1]); c.h([0,1])
    c.measure_all(); return sim.run(c,shots=1024).result().get_counts()

plot_histogram([m(qc,'Z'), m(qc,'X'), m(qc,'Y')], legend=['Z','X','Y'])
````

## Results & Discussion

### Results
- Z and X: only matching outcomes (00/11) dominate.
- Y: opposite outcomes (01/10) dominate.

### Discussion
- Stabilizer predictions verified by basis-dependent correlation patterns.
- Small off-pattern counts attributable to shot noise or transpilation details.
