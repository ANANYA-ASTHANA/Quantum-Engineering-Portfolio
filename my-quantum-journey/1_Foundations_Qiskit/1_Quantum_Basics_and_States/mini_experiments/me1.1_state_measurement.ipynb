# ME1.1 State Measurement of $|0\rangle, |1\rangle, |+\rangle, |-\rangle$

## Objectives
- Measure outcome distributions for $∣0\rangle,∣1\rangle,∣+\rangle,∣−\rangle$ in the Z basis.
- Verify Born-rule predictions with finite-shots (1024) sampling.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```


## Theory Snapshot
- $∣0\rangle,∣1\rangle$ are Z-eigenstates $\rightarrow$ deterministic outcomes.
- $∣\pm\rangle=(∣0\rangle\pm∣1\rangle)/\sqrt(2)$ $\rightarrow$ 50–50 in Z due to equal amplitudes.

## Experiment


````python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram

sim = Aer.get_backend('aer_simulator')

def run_and_measure(state):
    qc = state.copy()
    qc.measure_all()

    measure = sim.run(qc, shots=1024).result().get_counts()
    return measure

qc0 = QuantumCircuit(1) # |0>

qc1 = QuantumCircuit(1) # |1>
qc1.x(0)

qc_plus = QuantumCircuit(1) # |+>
qc_plus.h(0)

qc_minus = QuantumCircuit(1) # |->
qc_minus.h(0)
qc_minus.z(0)

c0 = run_and_measure(qc0)
c1 = run_and_measure(qc1)
c_plus = run_and_measure(qc_plus)
c_minus = run_and_measure(qc_minus)

plot_histogram([c0, c1, c_plus, c_minus], legend=["|0>", "|1>", "|+>", "|->"])
````

## Results & Discussion

### Results
- $∣0\rangle\rightarrow P(0)\approx 1, ∣1\rangle\rightarrow P(1)\approx 1$ (minor shot noise).
- $∣+\rangle,∣−\rangle\rightarrow P(0)\approx P(1)\approx 0.5$.

### Discussion
- Measurements match eigenstate vs superposition expectations.
- Deviations are within binomial shot noise for 1024 shots.
