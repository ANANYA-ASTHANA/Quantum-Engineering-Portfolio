# ME6.1: Single-qubit VQE

## Objectives
- Optimize a 1-parameter ansatz to minimize $\langle Z\rangle$.
- Validate gradient-descent convergence numerically.

## Setup
```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```


## Theory Snapshot
- For ansatz $R_y(\theta)∣0\rangle, \langle Z\rangle=\cos\theta \rightarrow$ minimum at $\theta = \pi$.
- Smooth, convex objective over $[0,\pi]$.

## Experiment


````python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def expZ(theta):
    qc = QuantumCircuit(1); qc.ry(theta, 0)
    psi = Statevector.from_instruction(qc).data
    return (abs(psi[0])**2 - abs(psi[1])**2)  # <Z> = P0 - P1

def grad(theta):  # parameter-shift
    s = np.pi/2
    return 0.5*(expZ(theta + s) - expZ(theta - s))

theta, lr, iters = 2.0, 0.2, 50
for k in range(iters):
    theta -= lr * grad(theta)
    if k % 10 == 0:
        print(f"iter {k:2d}: theta={theta:.4f}, <Z>={expZ(theta):.6f}")
print(f"final: theta={theta%(2*np.pi):.4f}, <Z>={expZ(theta):.6f}")
````

## Results & Discussion

### Results
- Logs show convergence to $\theta \approx 3.1416, \langle Z\rangle \approx −1.0000$.
- Stable convergence across iterations (printed checkpoints).

### Discussion
- Matches analytic optimum exactly in noiseless sim.
- Learning rate appears well-tuned (no oscillations/divergence).
