# ME6.2: 2-qubit VQE

## Objectives
- Evaluate $\langle ZZ\rangle$ for a 2-param ansatz across $(\theta_0,\theta_1)$.
- Identify parameter regions yielding strong (anti)correlations.

## Setup
```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```


## Theory Snapshot
- Product $R_y(\theta_0)\otimes R_y(\theta_1)$ gives $\langle ZZ\rangle=\cos\theta_0\cos\theta_1$ (no entanglers).
- Extremes near $\theta \in \{0,\pi\}$ maximize $∣\langle ZZ\rangle∣$.

## Experiment


````python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def ansatz(theta0, theta1):
    qc = QuantumCircuit(2)
    qc.ry(theta0, 0)
    qc.ry(theta1, 1)
    qc.cx(0, 1)
    return qc

def exp_ZZ(theta0, theta1):
    sv = Statevector.from_instruction(ansatz(theta0, theta1))
    p = sv.probabilities()  # order: 00,01,10,11
    return (+1)*p[0] + (-1)*p[1] + (-1)*p[2] + (+1)*p[3]

# demo sweep
for _ in range(10):
    t0 = np.random.uniform(0, 2*np.pi)
    t1 = np.random.uniform(0, 2*np.pi)
    print(f"θ0={t0:.3f}, θ1={t1:.3f} -> <ZZ> = {exp_ZZ(t0, t1):.6f}")
````

## Results & Discussion

### Results
- Printed samples include $\langle ZZ\rangle \approx −0.9966$ near certain $(\theta_0,\theta_1)$.
- Values vary smoothly; sign flips with quadrant (as cosine product).

### Discussion
- Observations align with separable model $\cos\theta_0\cos\theta_1$.
- Adding entanglers would enable richer objectives beyond factorized $ZZ$.
