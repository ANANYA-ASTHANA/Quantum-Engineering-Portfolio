## What We Learn
- How multi-qubit states are constructed using tensor products  
- How single-qubit gates extend to multi-qubit registers  
- How CX and CZ generate quantum entanglement  
- How to prepare Bell and GHZ states  
- How measurement in different bases reveals correlations

## Key Ideas

### 1) Tensor Products
Composite systems are built as: $|a\rangle \otimes |b\rangle$.  
A single-qubit gate acting on qubit *i* extends via identity on the rest.

### 2) Entanglement
A state is **entangled** if it **cannot** be written as: $|\psi\rangle = |u\rangle \otimes |v\rangle$.  
Entangled states exhibit correlations across bases.

### 3) Bell States
Example: $|\Phi^+\rangle = (|00\rangle+|11\rangle)/\sqrt{2}$.  
- Perfect correlation in Z and X  
- Characteristic relations in Y

### 4) GHZ States
Three-qubit generalization revealing multipartite nonclassical structure.  
Measurement of one qubit conditions the remaining pair.

## Example Code — Bell and GHZ

```python
from qiskit import QuantumCircuit

# Bell state
qc = QuantumCircuit(2)
qc.h(0); qc.cx(0,1)
qc.measure_all()
qc.draw("mpl")
```

```python
# GHZ state
from qiskit import QuantumCircuit
qc = QuantumCircuit(3)
qc.h(0); qc.cx(0,1); qc.cx(1,2)
qc.measure_all()
qc.draw("mpl")
```

## Summary
Entanglement arises from controlled operations and manifests through cross-basis measurement correlations. Bell and GHZ states provide canonical examples of this structure.
