# Module 3 — Quantum Gates & Circuits 

## What We Learn
- How Pauli, Hadamard, and phase gates affect qubits  
- How continuous rotations work  
- How entangling gates produce multi-qubit structure  
- How universal gate sets are formed

## Key Ideas

### 1) Pauli Gates
Act as reflections/rotations about Bloch axes.  
- X $ \rightarrow $ flips state  
- Y, Z $ \rightarrow $ rotate around axes

### 2) Hadamard (H)
- Creates superposition  
- Converts between Z and X bases

### 3) Phase Gates
S, T, and continuous $R_z(\theta)$:
- Modify relative phase  
- Affect interference  
- Leave Z-basis probabilities unchanged

### 4) Entangling Gates
CX and CZ correlate amplitudes and create entanglement.  
**Universal Set:** {H, T, CX}

## Example Code — Single- and Two-Qubit Patterns

```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(1)
qc.h(0); qc.t(0); qc.sdg(0)
qc.draw("mpl")
```

```python
# Entangling pattern
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0); qc.cx(0,1)
qc.draw("mpl")
```
