# ME3.2: $S, S^\dagger, T, T^\dagger$ Phase Demos

## Objectives
- Apply $S,S^\dagger,T,T^\dagger$ to $∣1\rangle$; inspect statevectors.
- Connect resultant phases to points on Bloch Sphere.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_multivector
from qiskit.quantum_info import Statevector
```


## Theory Snapshot
- $S : |1\rangle \mapsto\ i|1\rangle, \quad S^\dagger : |1\rangle \mapsto\ -i|1\rangle.$
- $T : |1\rangle \mapsto\ e^{i\pi/4}|1\rangle, \quad T^\dagger : |1\rangle \mapsto\ e^{-i\pi/4}|1\rangle.$

## Experiment


````python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_multivector
from qiskit.quantum_info import Statevector

gate_ops = {
    's':   lambda qc: qc.s(0),
    'sdg': lambda qc: qc.sdg(0),
    't':   lambda qc: qc.t(0),
    'tdg': lambda qc: qc.tdg(0),
}

for name, apply_gate in gate_ops.items():
    qc = QuantumCircuit(1)
    qc.x(0)            
    apply_gate(qc)  
    state = Statevector.from_instruction(qc)
    print(f"{name}: {state.data}")
    fig = plot_bloch_multivector(state)
    fig.suptitle(f"{name.upper()} applied to |1⟩")
    display(fig)
````

## Results & Discussion

### Results
- Prints show: $S \rightarrow [0,+i], S^\dagger \rightarrow [0,-i], T \rightarrow (1+i)/\sqrt(2), T^\dagger \rightarrow (1-i)/\sqrt(2)$.
- Amplitudes match expected global/relative phase factors.

### Discussion
- Phase gates rotate around Z; population unchanged in Z-basis.
- Differences become measurable under interference or non-Z readouts.
