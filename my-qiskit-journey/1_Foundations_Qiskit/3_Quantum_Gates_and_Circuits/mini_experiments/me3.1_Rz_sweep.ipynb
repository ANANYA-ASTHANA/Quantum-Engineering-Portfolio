# ME3.1: $R_z(\theta)$ sweep on $|+\rangle$

## Objectives
- Measure X-basis response to $R_z(\theta)$ using $H–R_z–H$.
- Observe interference vs $\theta$ and relate to $\langle X\rangle$ trends.

## Setup
```python
import numpy as np, matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import Aer
```


## Theory Snapshot
- $R_z(\theta)$ adds a relative phase; $H–R_z–H$ maps it to X-basis amplitudes.
- $P(+)=\frac{1+\cos\theta}{2}$ and $\langle X\rangle = P(+) - P(-) = \cos\theta$.

## Experiment



````python
import numpy as np, matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import Aer

sim = Aer.get_backend('aer_simulator')

def p_plus(theta):
    qc = QuantumCircuit(1)
    qc.h(0); qc.rz(theta,0); qc.h(0); qc.measure_all()
    result = sim.run(qc, shots=2000).result().get_counts()
    return result.get('0',0)/2000  # '0' = |+>

thetas = np.linspace(0, 4*np.pi, 80)  # 80 equal divisions from 0 to 4*pi
p_plus_vals = [p_plus(t) for t in thetas]

plt.plot(thetas, p_plus_vals)
plt.xlabel("θ (radians)")
plt.ylabel("P(+)")
plt.title("X-basis probability after Rz(θ) on |+⟩")
plt.show()
````

## Results & Discussion

### Results
- $P(+)$ follows $\frac{1+\cos\theta}{2}$ with clear X-basis fringes.
- Maxima near $\theta=0$, minima near $\theta=\pi$.

### Discussion
- $H–R_z–H$ converts phase into X-basis amplitudes, producing the observed $\cos\theta$ pattern.
- Minor deviations from ideal behavior arise from shot noise and transpilation.
