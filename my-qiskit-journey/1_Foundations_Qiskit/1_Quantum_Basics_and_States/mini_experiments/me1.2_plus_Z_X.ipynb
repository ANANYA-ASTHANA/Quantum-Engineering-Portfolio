# ME1.2: $|+\rangle$ in Z/X Bases

## Objectives
- Compare measuring $∣+\rangle$ in Z vs after Hadamard (X-basis readout).
- Show basis-change via H turns relative phase into deterministic outcome.

## Setup
```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```


## Theory Snapshot
- H maps X $\leftrightarrow$ Z: $∣+\rangle$ is +1 eigenstate of X.
- Z-basis on $∣+\rangle$ yields 50–50; X-basis yields deterministic “+”.

## Experiment



````python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram

qc=QuantumCircuit(1); qc.h(0)
sim=Aer.get_backend('aer_simulator')

def m(c,b='Z'):
    c=c.copy()
    if b=='X': c.h(0)
    c.measure_all(); return sim.run(c,shots=1024).result().get_counts()

plot_histogram([m(qc,'Z'), m(qc,'X')], legend=['Z','X'])
````

## Results & Discussion

### Results
- Z-basis: $\approx$ 0.5/0.5 split.
- X-basis (via H then Z): all counts in the “+” outcome.

### Discussion
- Basis alignment collapses superposition deterministically.
- Confirms role of H as a basis-rotation operator.
