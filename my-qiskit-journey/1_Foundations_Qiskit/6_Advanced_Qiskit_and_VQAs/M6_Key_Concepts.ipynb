## Key Ideas

### 1) Parameterized Circuits
**Ansatz:** $U(\vec{\theta}) = U(\theta_1, \theta_2, \dots)$. 
Used in VQE, QAOA, and general variational methods.

### 2) Expectation Values
$\langle H \rangle_{\theta} = \langle \psi(\theta) | H | \psi(\theta) \rangle$  
Computed efficiently using the **Estimator** primitive.

### 3) Optimization
- Gradient-free or gradient-based  
- Parameter-shift rules for analytic gradients  
- Cost landscapes can have local minima and barren plateaus

## Example Code — Estimator & 1‑Qubit Ansatz

```python
from qiskit import QuantumCircuit
from qiskit_aer.primitives import Estimator
import numpy as np

qc = QuantumCircuit(1)
qc.ry(np.pi/4, 0)

est = Estimator()
value = est.run(qc, ["Z"]).result().values[0]
value
```
